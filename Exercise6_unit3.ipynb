```python
#Operations in tuples
t1=(3,5,6,1,"ankit","kowshik")
t2=(7,)
print(t1+t2)
print(t2*3)
print(len(t2))
for i in t1:
  print(i)
if 3 in t1:
  print("yes")
```

```python
#Methods in tuples
print(t1.count(5))
print(t2.index(7))
```

```python
#Built in functions in tuples
t3=("hello","python","programming")
t4=(7439,34768,98787)
print(len(t4))
print(max(t4))
print(min(t4))
print(sum(t4))
print(sorted(t4))
print(t4.count(34768))
print(t4.index(98787))
```
