```python
#write a program to copy content from one file to another file
with open("f1.txt","r") as f:
  data =f.read()
print("reading done")
with open("f2.txt","w") as f:
  f.write(data)
print("data copied")
```

```python
#Write a program to print each line in reverse order
with open("f1.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```

```python
#write a program to sort words in a file,file should have only lower case words
with open("f1.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
  with open("f4.txt","w") as f:
    for i in sort_data:
      f.write(i+"\n")
  print("done")
```

```python
#Python program to compute the no.of chracters,words and lines in a file
with open("f1.txt","r") as f:
  data=f.read()
characters=len(data)
words=len(data.split())
lines=len(data.splitlines())
print(characters,words,lines,sep=",")
```

```python
#rb,wb,ab
import pickle
with open("s1.dat","wb") as f :
  data=[1,2,3,"hello"]
  pickle.dump(data,f)
with open("s1.dat","rb") as f:
  d=pickle.load(f)
  print(d)
```
