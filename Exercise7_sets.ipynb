```python
#set methods in python
a={1,2,56,78,"ankit","kumar"}
b={11,67,54,9,8,6}
a.add(5)
print(a)
a.update((6,7))
print(a)
a.remove(7)
print(a)
a.discard(6)
print(a)
a.pop()
print(a)
a.clear()
print(a)
a={1,2,3,4}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

```python
#PROGRAM to find no.of duplicates in a list
l=[int(i) for i in input().split()]
s=set(l)
print(s)
print(len(l)-len(s))  #no.of duplicates
```

```python
l=[int(i) for i in input().split()]
l1=[]
for i in l:
  if l.count(i)>1:
    l1.append(i)
print(l1)
s=set(l1)
print(s)
print(len(s))         #no.of numbers having duplicates
```

```python
l1=[int(i) for i in input().split()]
l2=[]
for i in l1:
  if l1.count(i)==1:
    l2.append(i)

print(l2)
s2=set(l2)
print(s2)          #print unique number
print(len(s2))      #no.of unique number


```

```python
l1=[int(i) for i in input().split()]
l2=[]
for i in l1:
  if l1.count(i)>1:
    l2.append(i)
s1=set(l1)
s2=set(l2)
print(s1-s2)
```

```python
#program that prints missing numbers between 1 to n
n= int(input("enter the number:"))
l=[int(i) for i in input().split()]
l1=set(l)
print(l1)
elements=set(range(1,n+1))
print(elements)
print(elements-l1)
```
