```python
#list operations
#concatenation
L1=[2,7.89,'hello','ankit']
L2=['hello','pavan']
print(L1+L2)
```

```python
#repetition
print(L2*3)
```

```python
#indexing
print(L1[3])
print(L1[-3])
```

```python
#slicing
L3=[1,89,7,'gok gok','harshith']
print(L3[0:5:2])
```

```python
#Membership Operators
if 10 not in L3:
  print("not in there")
if 7 in L3:
  print("it is there")
```

```python
#iterations
for i in L3:
  print(i,end=" ")
```

```python
#Built in Functions
print(len(L3))
L4=[3,7,5,1,9,-6]
print(sum(L4))
print(max(L4))
print(min(L4))
print(list(reversed(L4)))
print(sorted(L4))
```

```python
# list methods
L1=[1,2,3,'hai']
L1.append("python")
print(L1)
L3=[25,89,0]
L3.extend(L1)
print(L3)
L3.insert(0,"BSDK")
print(L3)
L3.pop(0)
print(L3)
L3.remove('hai')
L3.remove('python')
print(L3)
L3.reverse()
print(L3)
print(L3.index(3))
print(L3.count(1))
L3.sort()
print(L3)
```

```python
#List compression
L=[int(i) for i in input().split()]
print(L)
```
